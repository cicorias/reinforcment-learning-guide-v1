$\max 5x_1+4x_2+3x_3$ s.t. $2x_1+3x_2+x_3\le4$, $x\in\{0,1\}^3$. Guide: `02-intermediate/04`.

> **Predict before you run:** which node gets pruned by bound, and why?

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
from rlguide import branch_and_bound as bb
res = bb.branch_and_bound([5, 4, 3], [[2, 3, 1]], [4], [(0, 1)] * 3)
for i, node in enumerate(res.log):
    x = None if node.x is None else np.round(node.x, 3)
    print(f'node {i}: bounds={node.bounds}  LP={node.lp_value}  x={x}  -> {node.outcome}')
print('optimum', res.value, res.x)
```
