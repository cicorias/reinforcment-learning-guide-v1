# 08 · Simplex on the quiz LP
$\max 3x_1+2x_2$ s.t. $2x_1+x_2\le4$, $x_1+2x_2\le3$. Guide: `02-intermediate/04`.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
from rlguide import simplex
res = simplex.simplex([3, 2], [[2, 1], [1, 2]], [4, 3])
for enter, leave, ratios in res.pivots:
    print(f'{enter} enters, {leave} leaves (ratios {ratios})')
print('x* =', res.x, ' z* =', res.value)
```

```python
xs = np.linspace(0, 2.5, 200)
plt.fill([0, 2, 5/3, 0], [0, 0, 2/3, 1.5], alpha=.3, label='feasible polytope')
plt.plot(xs, 4 - 2*xs, label='2x1 + x2 = 4'); plt.plot(xs, (3 - xs)/2, label='x1 + 2x2 = 3')
plt.plot([0, 2, 5/3], [0, 0, 2/3], 'ko-', label='pivot path')
plt.plot(xs, (19/3 - 3*xs)/2, 'k--', label='z = 19/3')
plt.xlim(0, 2.5); plt.ylim(0, 2); plt.legend(); plt.gca().set_aspect('equal'); plt.show()
```
