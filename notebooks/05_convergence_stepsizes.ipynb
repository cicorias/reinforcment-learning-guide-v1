# 05 · Step sizes: constant α vs 1/n
Guide: `00-math-foundations/05`.

> **Predict before you run:** does the constant-α estimate ever stop jittering?

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
from rlguide import stochastic_approx as sa

@interact(alpha=FloatSlider(value=0.5, min=0.01, max=0.99, step=0.01),
          noise=FloatSlider(value=1.0, min=0.0, max=2.0, step=0.1))
def show(alpha, noise):
    rng = np.random.default_rng(0)
    x = 1.0 + noise * rng.normal(size=5000)
    plt.plot(sa.estimate_mean(x, alpha), lw=.6, label=f'constant alpha={alpha}')
    plt.plot(sa.estimate_mean(x, None), lw=2, label='alpha = 1/n')
    plt.axhline(1.0, color='k', ls='--'); plt.legend(); plt.show()
    print('theory: stationary variance =', sa.constant_alpha_stationary_variance(alpha, noise**2))
```
