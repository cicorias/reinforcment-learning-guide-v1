## Expected next-state value
$\mathbb E[V(S')] = \sum_{s'} P(s'\mid s)V(s')$

```python
p, V = np.array([0.2, 0.5, 0.3]), np.array([4, 1, 7])
print('expectation', p @ V, '| plain average', V.mean(), '| most likely', V[p.argmax()])
```

## Value iteration error vs the $\gamma^k$ bound
> **Predict before you run:** if you double $1-\gamma$, roughly how many fewer iterations do you need?

```python
from rlguide import envs, mdp_dp
P, R = envs.random_mdp(S=10, A=3, seed=0)

@interact(gamma=FloatSlider(value=0.9, min=0.5, max=0.99, step=0.01))
def show(gamma):
    v_star, _, _ = mdp_dp.value_iteration(P, R, gamma, theta=1e-12)
    V, errs = np.zeros(10), []
    for _ in range(200):
        V = mdp_dp.bellman_optimality(P, R, V, gamma)
        errs.append(mdp_dp.sup_norm(V - v_star))
    k = np.arange(1, 201)
    plt.semilogy(k, errs, label='||V_k - v*||')
    plt.semilogy(k, gamma**k * mdp_dp.sup_norm(v_star), '--', label='gamma^k ||V_0 - v*||')
    plt.xlabel('k'); plt.legend(); plt.show()
```
