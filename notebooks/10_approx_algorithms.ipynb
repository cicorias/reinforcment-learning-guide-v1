## MAX-SAT: derandomisation trace

```python
from rlguide import maxsat
C = maxsat.GUIDE_EXAMPLE
print('E[X] =', maxsat.expected_satisfied(C, {}))
assign, trace = maxsat.derandomized(C, 3)
for v, eT, eF, ch in trace:
    print(f'x{v}: E|T={eT:.3f}  E|F={eF:.3f}  -> {ch}')
print('satisfied:', maxsat.num_satisfied(C, assign), '/', len(C))
```

## Knapsack FPTAS: ε vs quality and table size

```python
from rlguide import knapsack_fptas as kf
rng = np.random.default_rng(0)
v = [int(x) for x in rng.integers(1, 1000, 25)]; w = [int(x) for x in rng.integers(1, 100, 25)]
W = int(sum(w) * .4); opt = kf.exact(v, w, W)[1]
for eps in [0.9, 0.5, 0.25, 0.1, 0.05]:
    _, val, info = kf.fptas(v, w, W, eps)
    print(f'eps={eps:<5} value/OPT={val/opt:.4f}  (guarantee >= {1-eps:.2f})  DP width={sum(info["scaled"])}')
```

## Vertex cover: LP rounding

```python
from rlguide import vertex_cover as vc
for name, n, E in [('path 1-2-3-4', 4, vc.PATH_1234), ('K5', 5, vc.complete_graph(5))]:
    C, lp = vc.lp_rounding(n, E)
    print(f'{name}: LP*={lp:.2f}  rounded |C|={len(C)}  OPT={vc.brute_force_opt(n, E)}')
```
