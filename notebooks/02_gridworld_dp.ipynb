# 02 · Dynamic programming: VI vs PI vs LP
Guide: `02-intermediate/01`, `/05`.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
from rlguide import envs, mdp_dp, mdp_lp
P, R = envs.random_mdp(S=12, A=4, seed=1)
v_vi, pi_vi, d_vi = mdp_dp.value_iteration(P, R, 0.9, theta=1e-12)
v_pi, pi_pi, iters = mdp_dp.policy_iteration(P, R, 0.9)
v_lp = mdp_lp.solve_primal(P, R, 0.9)
print('max |VI-PI| =', np.abs(v_vi-v_pi).max(), '  max |LP-PI| =', np.abs(v_lp-v_pi).max())
print('PI iterations:', iters, '  VI sweeps:', len(d_vi))
```

## Jacobi vs Gauss–Seidel evaluation
> **Predict before you run:** which curve reaches $10^{-8}$ first?

```python
pi = np.full(R.shape, 1 / R.shape[1])
_, dj = mdp_dp.iterative_policy_evaluation(P, R, pi, 0.9, theta=1e-8)
_, dg = mdp_dp.iterative_policy_evaluation(P, R, pi, 0.9, theta=1e-8, in_place=True)
plt.semilogy(dj, label=f'Jacobi ({len(dj)} sweeps)')
plt.semilogy(dg, label=f'Gauss-Seidel ({len(dg)} sweeps)')
plt.xlabel('sweep'); plt.ylabel('Delta_k'); plt.legend(); plt.show()
```
