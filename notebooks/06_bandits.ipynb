# 06 · Bandits: ε-greedy vs UCB vs Thompson
Guide: `01-beginner/05`.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
from rlguide import bandits
means = [0.2, 0.5, 1.0, 0.1, 0.8]

@interact(eps=FloatSlider(value=0.1, min=0.0, max=0.5, step=0.01),
          c=FloatSlider(value=2.0, min=0.1, max=4.0, step=0.1))
def show(eps, c):
    for label, kw in [(f'eps-greedy {eps}', dict(strategy='eps', eps=eps)),
                      ('greedy', dict(strategy='greedy')),
                      (f'UCB c={c}', dict(strategy='ucb', c=c))]:
        reg = np.mean([bandits.run_bandit(means, 3000, seed=s, **kw)['regret']
                       for s in range(20)], axis=0)
        plt.plot(reg, label=label)
    ts = np.mean([bandits.thompson_bernoulli([.2, .5, .7, .1, .6], 3000, seed=s)['regret']
                  for s in range(20)], axis=0)
    plt.plot(ts, label='Thompson (Bernoulli arms)')
    plt.ylabel('cumulative regret'); plt.legend(); plt.show()
```
