# 04 · Cliff walking: SARSA vs Expected SARSA vs Q-learning
Guide: `02-intermediate/02`.

> **Predict before you run:** which algorithm earns the best *online* return, and which has the shortest greedy path?

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
from rlguide import td_control as tc

@interact(eps=FloatSlider(value=0.1, min=0.0, max=0.3, step=0.02),
          alpha=FloatSlider(value=0.5, min=0.05, max=1.0, step=0.05))
def compare(eps, alpha):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 3.5))
    for m in ('sarsa', 'expected_sarsa', 'qlearning'):
        Q, ret = tc.run(m, episodes=500, alpha=alpha, eps=eps, seed=0)
        ax1.plot(np.convolve(ret, np.ones(20) / 20, 'valid'), label=m)
        path = tc.greedy_path(Q)
        if path:
            ys, xs = zip(*path)
            ax2.plot(xs, ys, marker='.', label=f'{m} ({len(path)-1} steps)')
    ax1.set_ylim(-120, 0); ax1.set_title('online return (20-ep avg)'); ax1.legend()
    ax2.fill_between([0.5, 10.5], 2.5, 3.5, color='k', alpha=.3)
    ax2.invert_yaxis(); ax2.set_title('greedy paths (black = cliff)'); ax2.legend()
    plt.show()
```
