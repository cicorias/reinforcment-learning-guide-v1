# 11 · Function approximation and the deadly triad
Guide: `03-expert/03`. Baird's counterexample: 7 states, linear features, off-policy semi-gradient TD(0) **diverges**.

> **Predict before you run:** what happens to $\lVert w\rVert$ as updates accumulate?

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, IntSlider
np.set_printoptions(precision=4, suppress=True)
```

```python
# Baird (1995), as in Sutton & Barto Example 11.1 (expected-update version)
X = np.zeros((7, 8))
for i in range(6):
    X[i, i] = 2; X[i, 7] = 1
X[6, 6] = 1; X[6, 7] = 2
gamma, alpha = 0.99, 0.01
w = np.array([1, 1, 1, 1, 1, 1, 10, 1], float)
norms = []
for _ in range(1000):
    # target policy always 'solid' -> state 7; behaviour visits states uniformly
    delta = gamma * X[6] @ w - X @ w          # reward is 0 everywhere
    w = w + alpha * (X.T @ delta) / 7
    norms.append(np.linalg.norm(w))
plt.semilogy(norms); plt.xlabel('sweep'); plt.ylabel('||w||'); plt.title('divergence'); plt.show()
```

Remove any one leg of the triad (tabular features, Monte Carlo targets, or on-policy state weighting) and the divergence disappears. DQN's replay buffer and target network are engineering answers to this instability.
